Aula - if, else e elif

```python
idade = int(input("Digite sua idade:"))

if idade < 16:
              print("Menor de Idade")

elif idade >= 16 and idade < 18:
              print("Voto Opcional")

elif idade >= 18:
              print("Voto Obrigatório")

else:
              print("Nenhuma das Opções")
```

```python
nota = float(input("Digite sua Nota:"))

if nota >= 7:
  print("Aprovado!")

elif nota >= 5 and nota < 7:
  print("Recuperação!")

elif nota < 5:
  print("Reprovado!")

else:
   print("Nota Invalida")
```

```python
cartao = input("Voçê possui o cartão da loja?(sim/não):")
valor = float(input("Qual o valor da sua compra?:"))
desconto = valor * 0.10
total = valor - desconto

if cartao != "sim" or valor < 100:
  print("Voçê não possui desconto!!")

elif cartao == "sim" and valor >= 100:
  print("Voçê recebeu 10% de desconto, o valor final é", total)

else:
   print("dados Invalidos")
```
